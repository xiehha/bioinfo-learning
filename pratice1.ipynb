```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([0.001,0.1,0.5,1,2,4,5,16,32,64,128,256,512,1000,10000])
y=np.log2(x)
plt.plot(x,y,marker='o',color='b')
plt.title('Raw Fold Change vs Log2 Fold Change')
plt.xlabel('Raw Fold Change (X)')
plt.ylabel('Log2 Fold Change(Y)')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
data = {
    '基因':['Gene_1','Gene_2','Gene_3','Gene_4','Gene_5'],
    '正常组':[10,50,100,20,5],
    '疾病组':[15,200,110,18,50]
}
df = pd.DataFrame(data)
df.head()
```

```python
mean_normal = df['正常组'].mean()
mean_disease = df['疾病组'].mean()
import numpy as np
log2fc = np.log2(mean_disease / mean_normal)
print(log2fc)
```

```python
df['log2FC'] = np.log2(df['疾病组'] / df['正常组'])
df.head()
```

```python
df[df['log2FC'] > 1]
```

```python
import numpy as np

# 假设这是四个基因的 P 值
p_values = [0.1, 0.01, 0.001, 0.0001]

for p in p_values:
    y_axis = -np.log10(p)  # 这就是 -log10 的代码写法
    print(f"P值是 {p}，画在火山图Y轴上的高度是: {y_axis}")
```
